# Answer 19

```python
def check_age(age):
    if age < 18:
        raise ValueError('Age must be 18 or older')
    return "Apki age eligible hai"

try:
    user=int(input("Entre the age"))
    result=check_age(user)
    print(result)

except ValueError:
    print('You are not eligible')
```
